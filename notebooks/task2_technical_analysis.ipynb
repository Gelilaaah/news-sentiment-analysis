## Quantitative Analysis with TA-Lib and PyNance

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import talib

plt.style.use("seaborn-v0_8")
```

```python
# 1. Import os
import os
import pandas as pd

# 2. Path to your data folder
DATA_FOLDER = "../data/stock-data"

# 3. Read all CSVs inside the folder
csv_files = [f for f in os.listdir(DATA_FOLDER) if f.endswith(".csv")]
print("CSV files found:", csv_files)

# 4. Load all CSVs into a dictionary of DataFrames
dfs = {}
for file in csv_files:
    file_path = os.path.join(DATA_FOLDER, file)
    dfs[file] = pd.read_csv(file_path)

# 5. Preview one of the CSVs
dfs[csv_files[0]].head()
```

```python
for name, df in dfs.items():
    print(f"=== {name} ===")
    display(df.head())
    print(df.info())
    print("\n")
    # displays the first few rows of the loaded csv dataframes
```

```python
for df in dfs.values():
    if 'Date' in df.columns:
        df['Date'] = pd.to_datetime(df['Date'])
        # Converts the 'Date' column in each DataFrame to datetime format for analysis.
```

```python
print(dfs.keys()) # Prints the names of all loaded CSV files stored as DataFrames.
```

## Prepare AMZN.csv


```python
df_stock = dfs['AMZN.csv']  # use the exact filename
df_stock['Date'] = pd.to_datetime(df_stock['Date'])  # ensure date is datetime
df_stock = df_stock.sort_values('Date')
df_stock.set_index('Date', inplace=True)
df_stock.head()
```

```python
for name, df in dfs.items():
    df['Date'] = pd.to_datetime(df['Date'])
    df.sort_values('Date', inplace=True)
    df.set_index('Date', inplace=True)
    print(f"Processed {name}:")
    display(df.head())  # process all stock data
```

```python
import talib

df['SMA_20'] = talib.SMA(df['Close'], timeperiod=20)

df['SMA_50'] = talib.SMA(df['Close'], timeperiod=50)
# Calculates 20-day and 50-day SMA for the 'Close' price using TA-Lib.
```

```python

df['EMA_20'] = talib.EMA(df['Close'], timeperiod=20)
df['EMA_50'] = talib.EMA(df['Close'], timeperiod=50)
# Calculates 20-day and 50-day EMA for the 'Close' price using TA-Lib.
```

```python
df['RSI_14'] = talib.RSI(df['Close'], timeperiod=14) # Calculates 14-day Relative Strength Index (RSI) for the 'Close' price using TA-Lib.
```

```python
df['MACD'], df['MACD_signal'], df['MACD_hist'] = talib.MACD(df['Close']) # Calculates MACD, MACD signal, and MACD histogram for the 'Close' price using TA-Lib.
```

```python
df['ATR_14'] = talib.ATR(df['High'], df['Low'], df['Close'], timeperiod=14) # Calculates 14-day Average True Range (ATR) using High, Low, and Close prices with TA-Lib
```

```python
import talib

for name, df in dfs.items():
    df['SMA_20'] = talib.SMA(df['Close'], timeperiod=20)
    df['EMA_20'] = talib.EMA(df['Close'], timeperiod=20)
    df['RSI_14'] = talib.RSI(df['Close'], timeperiod=14)
    df['MACD'], df['MACD_signal'], df['MACD_hist'] = talib.MACD(df['Close'])
    df['BB_upper'], df['BB_middle'], df['BB_lower'] = talib.BBANDS(df['Close'])
    df['ATR_14'] = talib.ATR(df['High'], df['Low'], df['Close'], timeperiod=14)
    print(f"Indicators calculated for {name}")
    # Calculates multiple TA-Lib indicators (SMA, EMA, RSI, MACD, Bollinger Bands, ATR) for all loaded stock DataFrames.
```

## Generate RSI Trading Signals
Creates buy/sell signals based on RSI:

- RSI > 70 → Sell (-1)
- RSI < 30 → Buy (1) Prints signal counts for each stock.


```python
for stock_name, df in dfs.items():
    df['RSI_signal'] = 0
    if 'RSI_14' in df.columns:
        df.loc[df['RSI_14'] > 70, 'RSI_signal'] = -1  # Sell
        df.loc[df['RSI_14'] < 30, 'RSI_signal'] = 1   # Buy
    print(f"{stock_name} RSI signals:\n", df['RSI_signal'].value_counts(), "\n")
```

```python
# Computes and prints the correlation matrix of all numeric columns for each stock.

for stock_name, df in dfs.items():
    print(f"--- {stock_name} ---")
    
    # Only keep numeric columns
    numeric_cols = df.select_dtypes(include='number').columns
    corr_matrix = df[numeric_cols].corr()
    
    print(corr_matrix, "\n")
```

# Company-Specific Insights

## 1. Apple (AAPL)

### Trend Indicators

* The correlation between **Close price and SMA_20 (0.998)** and **EMA_20 (0.999)** is extremely strong.
* This confirms that Apple followed a **stable long-term upward trend** during the analyzed period.

### Momentum Indicators

* RSI has a weak negative correlation with price (-0.021), suggesting that momentum changes do not immediately affect price movement.
* MACD shows a moderate positive relationship with price (0.256), meaning bullish momentum often aligns with rising prices.

### Volatility

* ATR_14 has a strong positive relationship with price (0.91), indicating higher prices were accompanied by increased market volatility.

### Insight

Apple demonstrates:

* Stable trend behavior
* Strong moving-average alignment
* Moderate momentum signals
* Relatively controlled volatility

This makes AAPL suitable for:

* Trend-following strategies
* Moving-average crossover systems
